# Mining Refubium

*Refubium* basiert auf [DSpace](https://dspace.org/) und verfügt über eine OAI-Schnittstelle (vgl. [FU-Seite](https://www.fu-berlin.de/sites/ub/publizieren/refubium/ueber-uns/index.html)).

Kurzdokumentation der Schnittstelle unter <https://www.fu-berlin.de/sites/ub/publizieren/refubium/nutzen/index.html>.

## Daten abziehen

Mit der [`scythe`-Bibliothek](https://github.com/afuetterer/oaipmh-scythe) ist der OAI-Abruf in Python recht bequem möglich.

```python
from oaipmh_scythe import Scythe
from random import choice

repo_url = "https://refubium.fu-berlin.de/oai/request"
```

### Welche Sets gibt es?

```python
with Scythe(repo_url) as scythe:
    sets = scythe.list_sets()
    setList = [ s for s in sets ]
```

```python
for n,s in enumerate(setList):
    print(f"{n}\t{s.setName}")
```

### Daten eines Sets beziehen

```python
random_set = choice(setList) # wählt ein zufälliges Set aus
print("Gewähltes Set:", random_set.setName)
```

```python
with Scythe(repo_url) as scythe:
    try:
        records = scythe.list_records(set_=random_set.setSpec)
        recordList = [ r for r in records ] 
        print(f"{len(recordList)} Records im Set {random_set.setName}")
    except Exception as e:
        print(e)
```

## Daten analysieren

```python
from collections import Counter
```

### Rechteangaben

```python
rightsList = list()
for r in recordList:
    rights = r.get_metadata().get('rights', [])
    if len(rights) == 1:
        rightsList.append(rights[0])
```

```python
Counter(rightsList).most_common()
```

### Sprache

```python
langList = list()
for r in recordList:
    langs = r.get_metadata().get('language', [])
    if len(langs) == 1:
        langList.append(tuple(sorted(langs)))
```

```python
Counter(langList)
```

### Subjects

```python
subjectList = list()
for r in recordList:
    subjects = r.get_metadata().get('subject', [])
    if len(subjects) == 1:
        subjectList.append(tuple(sorted(subjects)))
```

```python
Counter(subjectList)
```

```python
for s in subjectList:
    print(s)
```

### Informationen zu einem Objekt abrufen

Für den Abruf wir der OAI-Identifier benötigt, der über das Web-Interface nicht explizit exponiert wird.
Er lässt sich jedoch aus dem Handle ableiten:

Aus `https://refubium.fu-berlin.de/handle/fub188/14889` ergibt sich der OAI-Identifier `oai:refubium.fu-berlin.de:fub188/14889`.

```python
with Scythe(repo_url) as scythe:
    record = scythe.get_record(identifier = "oai:refubium.fu-berlin.de:fub188/14889")
    print(record)
```

### Zu welchen Sets gehört ein Record?

```python
with Scythe(repo_url) as scythe:
    record = scythe.get_record(identifier = "oai:refubium.fu-berlin.de:fub188/14889")
    #print(record.header.setSpecs)

    # Informationen zu Listen abholen
    sets = { s.setSpec : s for s in scythe.list_sets() }
    for r in record.header.setSpecs:
        print(f"{r}\t{sets.get(r).setName}\t{sets.get(r).setDescription}")
        
```

## Dateitypen von Forschungsdaten

```python
formats = list()
with Scythe(repo_url) as scythe:
    records = list(scythe.list_records(set_="col_fub188_7"))
    for r in records:
        formats.extend(r.metadata.get('format', []))
```

```python
Counter(formats).most_common()
```
